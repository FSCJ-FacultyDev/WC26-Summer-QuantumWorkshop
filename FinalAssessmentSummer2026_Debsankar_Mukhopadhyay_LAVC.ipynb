```python
# Debsankar Mukhopadhyay
# Grover Search with Two Targets: k = 3 and k = 6

# Step 1: Define target predicate
def is_target(k):
    return k in [3, 6]


# Step 2: Define phase oracle
def oracle(state):

    result = state.copy()

    for k in range(len(result)):
        if is_target(k):
            result[k] *= -1

    return result


# Step 3: Define diffusion operator
def diffusion(state):

    mean = sum(state) / len(state)

    result = state.copy()

    for k in range(len(result)):
        result[k] = 2 * mean - result[k]

    return result


# Step 4: Initialize 3-qubit circuit and create uniform superposition
qr = QuantumRegister(3)
qc = QuantumCircuit(qr)

qc.h(0)
qc.h(1)
qc.h(2)


# Step 5: Run circuit and display initial state table
state = qc.run()

print("INITIAL STATE")
print_state_table(state)


# Step 6: Copy original state
original_state = state.copy()


# Step 7: Apply oracle and display result
oracle_state = oracle(state)

print("\nAFTER ORACLE")
print_state_table(oracle_state)


# Step 8: Apply one Grover iteration (oracle + diffusion)
grover_state = diffusion(oracle_state)

print("\nAFTER ONE GROVER ITERATION")
print_state_table(grover_state)


# Step 9: Measure final state over 1000 shots
print("\nMEASUREMENT RESULTS (1000 SHOTS)")

counts = measure(grover_state, 1000)

print(counts)
```

**What has been done?**
In this exercise, Grover's Search Algorithm was implemented on a 3-qubit system to identify two marked states, 3 (011) and 6 (110), from the eight possible outcomes.

1. Create a Uniform Superposition
A quantum register containing 3 qubits was created using the QuantumRegister and QuantumCircuit classes. Hadamard gates were applied to all three qubits:
qc.h(0)
qc.h(1)
qc.h(2)

This placed the system into a uniform superposition, where each of the 8 possible states had the same amplitude and probability of being measured.

2. Apply the Oracle
A target predicate was defined to identify the marked states:

def is_target(k):
    return k in [3, 6]

The oracle then flipped the sign of the amplitudes corresponding to states 011 and 110. This phase flip marks the desired states without changing their probabilities.

3. Apply the Diffusion Operator
The diffusion operator performed an inversion about the mean:

result[k] = 2 * mean - result[k]

This step increased the amplitudes of the marked states and decreased the amplitudes of the unmarked states. As a result, the probability of measuring the target states was amplified.

4. Perform One Grover Iteration
A single Grover iteration consisted of:

Oracle → Diffusion

After the iteration, most of the probability amplitude was concentrated on the target states 011 and 110, making them much more likely to be observed when measured.

5. Measure the Final State
The final state was measured 1000 times using:

counts = measure(grover_state, 1000)

The measurement results showed that states 011 and 110 occurred most frequently, demonstrating that Grover's algorithm successfully amplified the probability of the marked outcomes.

**Conclusion**

This experiment demonstrated how Grover's Search Algorithm can efficiently locate multiple target states in a search space. Starting from an equal superposition of all eight possible 3-qubit states, the oracle marked the desired outcomes (3 and 6), and the diffusion operator amplified their amplitudes. After one Grover iteration, measurements overwhelmingly returned the marked states, confirming the effectiveness of quantum amplitude amplification.
